```python
# Creation of a tuple
a = (3, 4, 5, 6, 7)
print(a)
```

```python
# Reverse a tuple
a = (4, 5, 6, 7, 8)
print(a[::-1])
```

```python
# Remove duplicates from a tuple
a = (55, 66, 78, 44, 55)
print(tuple(set(a)))
```

```python
# Adding a mutable object to a tuple
b = (3, 4, [8, 9], 2)
b[2].append(10)
print(b)
```

```python
# To change elements of tuple
a = (4, 5, 6, 7)
a = a[:1] + (9,) + a[2:]
print(a)
```

```python
# Tuple slicing
a = (2, 4, 6, 7, 8)
print(a[0::2])
print(a[-5:])
```

```python
# Count the no. of elements in the tuple
t = (1, 2, 3, 4, 5, 6, 7, 2, 2)
c = t.count(2)
print(c)
```

```python
# Tuple unpacking
data = (109, "Ashifa", 19)
rollno, name, age = data
print(rollno, name, age)
```

```python
# List & Tuple Datatype transformation
l = [1, 2, 3]
t = (4, 5, 6)
print(l, t)
print(list(t), tuple(l))
```
