```python
# Creating a dictionary
d1 = {1: "apple", 2: "orange"}
print(d1)

d2 = {'name': 'gool', 'age': 19, 'qualification': 'AIML'}
print(d2)
```

```python
# Accessing a value in a dictionary
d2 = {'name': 'gool', 'age': 19, 'qualification': 'AIML'}
print(d2["name"])
print(d2["age"])
```

```python
# Adding a new key-value pair
d1 = {'name': 'gool', 'age': 19, 'dept': 'AIML'}
d1['city'] = 'Coimbatore'
print(d1)
```

```python
# Updating an existing value
d2 = {'name': 'gool', 'age': 19, 'qualification': 'AIML'}
d2["age"] = 20
print(d2)
```

```python
# Deleting a key-value pair
d2 = {'name': 'gool', 'age': 20, 'qualification': 'AIML'}
del d2["age"]
print(d2)
d2 = {'name': 'gool', 'age': 20, 'qualification': 'AIML'}
d2.pop("qualification")
print(d2)
```

```python
# Displaying all Keys, Values and key-value pairs
d2 = {'name': 'gool', 'age': 19, 'qualification': 'AIML'}
print(d2.keys())
print(d2.values())
print(d2.items())
```

```python
# Traversing the dictionary
d3 = {'name': 'gool', 'age': 19, 'gender': 'female', 'qualification': 'AIML'}
for i, j in d3.items():
    print(i, ":", j)
```
