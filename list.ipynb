```python
# Creating a list
a = ["yaal", 19, 109, "BSc AIML"]
print(a)
```

```python
# Added Elements to the list
a = ["yall", 19, 109, "BSc AIML"]
a.append("Shauthi")
print(a)
```

```python
# Removing an element from the list
a = ["Asifa", 19, 109, "BSc AIML"]
a.remove(19)
print(a)
```

```python
# Adding another list inside a list
a = ["yall", 19, 109, "BSc AIML"]
a.append([1, 2, 3])
print(a)
```

```python
a = [10, 20, 30, 70]
a.append([80, 60])
print(a[4])
```

```python
a = ["yall", 19, 109, "BSc AIML"]
a.extend("Shauthi")
print(a)
```

```python
# Reverse a list
a = ["yall", 19, 109, "BSc AIML"]
print(a[::-1])
```

```python
# Remove the duplicate elements from the list
a = [2, 3, 5, 5, 6, 7]
print(list(set(a)))
```

```python
a = [34, 56, 67, 27]
a.sort()
print(a[-2])
```

```python
# Find the second largest element in the list
a = [2, 3, 4, 5, 6, 7]
b = max(a)
a.remove(b)
print(max(a))
```

```python
# Find the common element in the list
a = [2, 3, 4, 5, 6]
b = [2, 5, 6, 9, 10]
for i in a:
    for j in b:
        if i == j:
            print(i)
```

```python
# Sorting without using sort function
l1 = [5, 6, 9, 1, 2]
for i in range(len(l1)):
    for j in range(i+1, len(l1)):
        if l1[i] > l1[j]:
            l1[i], l1[j] = l1[j], l1[i]
print(l1)
```
