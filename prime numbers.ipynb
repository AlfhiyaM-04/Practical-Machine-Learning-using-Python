```python
# (i) To check prime number or not
n = int(input())
c = 0
for i in range(1, n+1):
    if n % i == 0:
        c += 1

if c == 2:
    print(n, "It is a prime number")
else:
    print(n, "It is not a prime number")
```

```python
# (ii) To display prime numbers from 1 to 100 (using for)
for i in range(1, 101):
    c = 0
    for j in range(1, i+1):
        if i % j == 0:
            c += 1
    if c == 2:
        print(i)
```

```python
# (iii) To display prime numbers from 1 to 100 (using while)
i = 1
while i < 100:
    i += 1
    c = 0
    for j in range(1, i+1):
        if i % j == 0:
            c += 1
    if c == 2:
        print(i)
```
