```python
# User-Defined functions for string operations
def string_length(s):
    return len(s)

def to_upper(s):
    return s.upper()

def to_lower(s):
    return s.lower()

def reverse_string(s):
    return s[::-1]

text = input("Enter a string: ")
print("Length of the string:", string_length(text))
print("Uppercase:", to_upper(text))
print("Lowercase:", to_lower(text))
print("Reversed string:", reverse_string(text))
```
